```python
import numpy as np
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.models import Sequential, clone_model
from tensorflow.keras.optimizers import SGD

# 1. Define dataset (XOR)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# 2. Build model
model = Sequential([
    Input(shape=(2,)),
    Dense(16, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer=SGD(learning_rate=0.2),
    metrics=['accuracy']
)

# Clone model so Stochastic GD starts with fresh weights
model_sgd = clone_model(model)
model_sgd.compile(
    loss='binary_crossentropy',
    optimizer=SGD(learning_rate=0.1),
    metrics=['accuracy']
)

# --- 1. Batch Gradient Descent ---
model.fit(
    X, y,
    epochs=300,
    batch_size=len(X),
    verbose=0
)
loss_b, acc_b = model.evaluate(X, y, verbose=0)
print(f"Batch GD (batch_size={len(X)}) -> Accuracy: {acc_b * 100:.2f}%")

# --- 2. Stochastic Gradient Descent ---
model_sgd.fit(
    X, y,
    epochs=300,
    batch_size=1,  # 1 for True Stochastic GD
    verbose=0
)
loss_s, acc_s = model_sgd.evaluate(X, y, verbose=0)
print(f"Stochastic GD (batch_size=1)   -> Accuracy: {acc_s * 100:.2f}%")
```
